# 51. 결측치 비율

```python
missing_summaries = []

for name, df in datasets.items():

    summary = pd.DataFrame(

        {

            "missing_count": df.isna().sum(),

            "missing_rate_pct": (

                df.isna().mean() * 100

            ).round(2),

        }

    )

    summary.insert(0, "dataset", name)

    summary.insert(1, "column", summary.index)

    missing_summaries.append(summary.reset_index(drop=True))

missing_summary = pd.concat(

    missing_summaries,

    ignore_index=True

)

missing_summary
```

# Part 15. 파일 간 연결 무결성·반복 점검 함수·LLM 구조 요약

 

## 59. 참조하지 못한 고객 ID

```python

unknown_customer_ids = sorted(

    set(orders["customer_id"].dropna())
    - set(customers["customer_id"].dropna())

)

print("존재하지 않는 customer_id:", unknown_customer_ids)
```

## 60. 참조하지 못한 주문 ID

```python
unknown_order_ids = sorted(

    set(order_items["order_id"].dropna())

    - set(orders["order_id"].dropna())

)

print("존재하지 않는 order_id:", unknown_order_ids)
```

## 61. 참조하지 못한 상품 ID

```python
unknown_product_ids = sorted(

    set(order_items["product_id"].dropna())

    - set(products["product_id"].dropna())

)

print("존재하지 않는 product_id:", unknown_product_ids)
```

## 62. 반복 점검 함수

```python
def inspect_dataframe(

    name: str,

    df: pd.DataFrame,

    primary_key: str | None = None,

) -> dict:

    result = {

        "dataset": name,

        "rows": df.shape[0],

        "columns": df.shape[1],

        "column_names": df.columns.tolist(),

        "dtypes": {

            column: str(dtype)

            for column, dtype in df.dtypes.items()

        },

        "missing_total": int(df.isna().sum().sum()),

        "duplicate_rows": int(df.duplicated().sum()),

    }

    if primary_key is not None:

        result["primary_key"] = primary_key

        result["key_missing"] = int(

            df[primary_key].isna().sum()

        )

        result["key_duplicates"] = int(

            df[primary_key].duplicated().sum()

        )

        result["key_unique"] = int(

            df[primary_key].nunique()

        )

    return result
```

```python

inspection_results = []

for name, df in datasets.items():

    inspection_results.append(

        inspect_dataframe(

            name,

            df,

            primary_keys[name],

        )

    )

inspection_results
```

## 63. LLM에 전달할 안전한 구조 요약

```python

safe_schema_summary = {

    "datasets": inspection_results,

    "relationships": {

        "orders.customer_id": {

            "references": "customers.customer_id",

            "unknown_count": len(unknown_customer_ids),

        },

        "order_items.order_id": {

            "references": "orders.order_id",

            "unknown_count": len(unknown_order_ids),

        },

        "order_items.product_id": {

            "references": "products.product_id",

            "unknown_count": len(unknown_product_ids),

        },

    },

}

safe_schema_summary
```
